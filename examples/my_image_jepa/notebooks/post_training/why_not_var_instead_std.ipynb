```python
import torch 
import torch.nn as nn 
import torch.nn.functional as F 
```

```python
torch.manual_seed(123)

f1 = torch.randn(1,1)
f2 = torch.randn(1,1)
projector = nn.Linear(1,5,bias=False)
optimizer = torch.optim.SGD(projector.parameters(), lr = 0.01)

print(f'f1 : {f1}')
print(f'f2 : {f2}')
```

```python
for i in range(100): 
    print(f'epoch {i}')
    optimizer.zero_grad()

    z1 = projector(f1)
    z2 = projector(f2)
    print(f'z1 : {z1}')
    print(f'z2 : {z2}')
    print(f'proj : {projector.weight.data.T}')
    print('---'*10)

    sim_loss = F.mse_loss(z1,z2)
    Hinge_loss_z1 = F.relu(1- z1.std())
    Hinge_loss_z2 = F.relu(1- z2.std())
    loss = sim_loss + Hinge_loss_z1 + Hinge_loss_z2

    print(f'sim loss : {sim_loss.item():.4f}')
    print(f'Hinge_loss_z1 : {Hinge_loss_z1:.4f}')
    print(f'Hinge_loss_z2 : {Hinge_loss_z2:.4f}')
    print(f'loss {loss.item():.4f}')
    print("===="*30)
    
    loss.backward()
    optimizer.step()
        
```
