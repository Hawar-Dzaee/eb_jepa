```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
x   = np.linspace(-np.pi, np.pi, 100)
```

\[
f(x)=
\begin{cases}
-1, & -\pi < x < 0,\\
1, & 0 < x < \pi.
\end{cases}
\]

```python
def square(x):
    return np.where(x < 0, -1.0, 1.0)   # -1 on (-π,0), +1 on (0,π)

plt.plot(x, square(x), label="exact f(x)")
# plt.plot(x, sum(map(f, range(1, 1000))), label="Fourier (999 terms)")
plt.legend()
```

```python
L = 1 * np.pi
def f(n): 
   
    frac = 1 / L
    a = 2 / n
    c = (2 * np.cos(n * np.pi)) / n
    bn = frac * ( a - c)
    return bn * np.sin(n * x)
```

```python
y = f(1)
plt.plot(x,y)
```
