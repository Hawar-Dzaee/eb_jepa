```python
import torch 
import torch.nn as nn 
```

```python
class MyModule(nn.Module):
    def __init__(self,): 
        super().__init__()
        self.layers = nn.ModuleList(
            [nn.Linear(2,4) for _ in range(1)]
        )
        self.dp = nn.Dropout(p=0.5)

    def forward(self,x):
        for layer in self.layers:
            x = layer(x)
        return self.dp(x )
model = MyModule()
model
```

```python
output = model(torch.randn(2,2))
print(output.shape)
```
