```python
import torch 
import torch.nn as nn 
import torch.nn.functional as F 
```

```python


D_OBS = 1    # input channels (grayscale)
H_ENC = 32   # hidden dim in encoder
D_STC = 16   # representation dim (encoder output channels)
H_PRE = 32   # hidden dim in predictor

encoder_output = torch.randn(32,D_STC,10,64,64)
```

```python
# dataptive avg pooling 

x = [F.adaptive_avg_pool2d(encoder_output[:, :, t], (8,8)) for t in range(10)]
print(len(x))
print(x[0].shape)
```

```python
x = torch.stack(x,dim = 2)
print(x.shape)
```

```python
conv1  = nn.Conv3d(D_STC,H_PRE, kernel_size=(1,1,1), stride=(1,1,1),padding="same")
conv1_output = conv1(x)
print(conv1_output.shape)
```

```python
conv2 = nn.Conv3d(H_PRE, D_OBS,kernel_size=(1,3,3),stride=(1,1,1),padding="same")
conv2_output = conv2(conv1_output)
print(conv2_output.shape)
```

```python
output = conv2_output.squeeze(1)
print(output.shape)
```
