```python
import torch 
```

```python
torch.manual_seed(123)
state, pred = torch.randint(0,10, (2,1,4,2,2), dtype= torch.float32) , torch.randint(0,10, (2,1,4,2,2), dtype= torch.float16)

print(state)
```

```python
state_flatten, pred_flatten = state.transpose(0,1).flatten(1).transpose(0,1) , pred.transpose(0,1).flatten(1).transpose(0,1)
print(state_flatten.shape)
print(pred_flatten)
```

```python
import torch.nn.functional as F 


F.mse_loss(state_flatten, pred_flatten)
```
